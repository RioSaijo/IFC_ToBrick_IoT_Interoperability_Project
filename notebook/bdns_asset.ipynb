```python

# A と B の入力パスを指定
path_A = Path(r"C:\Users\Owner\Documents\25952501_RioSaijo\IoT-Interoperability-Project\data\intermediate\stage1_ir\bdns_assets.jsonl")
path_B = Path(r"C:\Users\Owner\Documents\25952501_RioSaijo\IoT-Interoperability-Project\resources\bdns\bdns_to_brick_mapping_candidates.csv")
```

```python

# 出力パスの既定値
out_dir = path_A.parent
out_csv  = out_dir / "bdns_assets_to_brick_map.csv"
out_xlsx = out_dir / "bdns_assets_to_brick_map.xlsx"

print(path_A)
print(path_B)
print(out_csv)
print(out_xlsx)
```

```python
def read_jsonl_records(fp: Path):
    """JSON Lines を逐次読込して dict のリストを返す"""
    records = []
    with fp.open("r", encoding="utf-8-sig") as f:
        for ln, line in enumerate(f, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                obj = json.loads(line)
            except json.JSONDecodeError as e:
                raise ValueError(f"JSONL 読込エラー 行={ln} . {e}") from e
            records.append(obj)
    return records


def to_abbr(name: str) -> str:
    """'AAA - BBB' → 'AAA' を返す. ハイフンの異体字も簡易対応"""
    if name is None:
        return ""
    s = str(name).strip()
    if not s:
        return ""
    # ' - ' を優先. 無い場合は最初に現れるハイフン類で分割
    if " - " in s:
        return s.split(" - ", 1)[0].strip()
    # 異体字ハイフン: -, ‐, ‑, ‒, –, —, −
    return re.split(r"\s*[-‐‑‒–—−]\s*", s, maxsplit=1)[0].strip()


def normalize_tag(s: str) -> str:
    """結合用キーの正規化: 前後空白除去 + 小文字化"""
    if s is None:
        return ""
    return str(s).strip().lower()
```

```python
# A を読込
recA = read_jsonl_records(path_A)

# 必要列の抽出
dfA = pd.DataFrame([{
    "name": r.get("name", ""),
    "bdns_tag": r.get("bdns_tag", ""),
    "raw_ifc_class": r.get("raw_ifc_class", "")
} for r in recA])

# 型と空白整形
dfA["name"] = dfA["name"].astype(str).str.strip()
dfA["bdns_tag"] = dfA["bdns_tag"].astype(str).str.strip()
dfA["raw_ifc_class"] = dfA["raw_ifc_class"].astype(str).str.strip()

# name → bdns_abbriviation へ変換
dfA["bdns_abbriviation"] = dfA["name"].map(to_abbr)
dfA = dfA.drop(columns=["name"])

# 結合キーを作成
dfA["_join_key"] = dfA["bdns_tag"].map(normalize_tag)

print(dfA.shape)
dfA.head()
```

```python
# B を読込
dfB = pd.read_csv(path_B, encoding="utf-8-sig")

# 列名の互換対応
# 既作成のファイルによっては 'input_tag' 列名の可能性があるため補正
if "bdns_tag" not in dfB.columns and "input_tag" in dfB.columns:
    dfB = dfB.rename(columns={"input_tag": "bdns_tag"})

# 必要列の存在検査
required_cols_B = {"bdns_tag", "brick_class_candidate"}
missing = [c for c in required_cols_B if c not in dfB.columns]
if missing:
    raise ValueError(f"B に必要列が存在しない . 欠落列={missing}")

# 正規化
dfB["bdns_tag"] = dfB["bdns_tag"].astype(str).str.strip()
dfB["brick_class_candidate"] = dfB["brick_class_candidate"].astype(str).str.strip()

# 重複タグは先勝ちで集約
dfB = dfB.drop_duplicates(subset=["bdns_tag"], keep="first").copy()

# 結合キーを作成
dfB["_join_key"] = dfB["bdns_tag"].map(normalize_tag)

print(dfB.shape)
dfB.head()
```

```python
# A を基準とした左外部結合
df = dfA.merge(
    dfB[["_join_key", "brick_class_candidate"]],
    on="_join_key",
    how="left"
)

# 最終列順
df_final = df[[
    "bdns_abbriviation",
    "bdns_tag",
    "raw_ifc_class",
    "brick_class_candidate",
]]

# 並びを安定化
df_final = df_final.sort_values(["bdns_abbriviation", "bdns_tag"], kind="stable").reset_index(drop=True)

# 概況
print("総レコード数 .", len(df_final))
unmatched = df_final["brick_class_candidate"].isna().sum()
print("Brick 未マッチ件数 .", unmatched)

df_final.head(10)
```

```python
# =========================
# 0) セットアップ
# =========================
from pathlib import Path
import json
import re
import pandas as pd

# 入力パス
path_A = Path(r"C:\Users\Owner\Documents\25952501_RioSaijo\IoT-Interoperability-Project\data\intermediate\stage1_ir\bdns_assets.jsonl")
path_B = Path(r"C:\Users\Owner\Documents\25952501_RioSaijo\IoT-Interoperability-Project\resources\bdns\bdns_to_brick_mapping_candidates.csv")

# 出力パス
out_dir  = path_A.parent
out_csv  = out_dir / "bdns_assets_to_brick_map.csv"
out_xlsx = out_dir / "bdns_assets_to_brick_map.xlsx"

print(f"A: {path_A}")
print(f"B: {path_B}")
print(f"OUT CSV : {out_csv}")
print(f"OUT XLSX: {out_xlsx}")
```

```python
# =========================
# 1) ユーティリティ
# =========================
def read_jsonl_records(fp: Path):
    """JSON Lines を読み込み dict のリストを返す"""
    records = []
    with fp.open("r", encoding="utf-8-sig") as f:
        for ln, line in enumerate(f, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                obj = json.loads(line)
            except json.JSONDecodeError as e:
                raise ValueError(f"JSONL 読込エラー 行={ln} , {e}") from e
            records.append(obj)
    return records

def to_abbr(name: str) -> str:
    """'AAA - BBB' → 'AAA' を返す, ハイフン異体字も簡易対応"""
    if name is None:
        return ""
    s = str(name).strip()
    if not s:
        return ""
    if " - " in s:
        return s.split(" - ", 1)[0].strip()
    # -, ‐, ‑, ‒, –, —, − に対応
    return re.split(r"\s*[-‐‑‒–—−]\s*", s, maxsplit=1)[0].strip()

def normalize_tag(s: str) -> str:
    """結合キー正規化, 前後空白除去 + 小文字化"""
    if s is None:
        return ""
    return str(s).strip().lower()
```

```python
# =========================
# 2) A を読込して整形
# =========================
recA = read_jsonl_records(path_A)

dfA = pd.DataFrame([{
    "name": r.get("name", ""),
    "bdns_tag": r.get("bdns_tag", ""),
    "raw_ifc_class": r.get("raw_ifc_class", "")
} for r in recA])

# 文字列正規化
dfA["name"] = dfA["name"].astype(str).str.strip()
dfA["bdns_tag"] = dfA["bdns_tag"].astype(str).str.strip()
dfA["raw_ifc_class"] = dfA["raw_ifc_class"].astype(str).str.strip()

# name → bdns_abbriviation へ変換
dfA["bdns_abbriviation"] = dfA["name"].map(to_abbr)
dfA = dfA.drop(columns=["name"])

# 結合キー
dfA["_join_key"] = dfA["bdns_tag"].map(normalize_tag)

# A 側 重複削除
# bdns_tag が同一の重複を先勝ちで 1 件に統合
dfA = dfA.drop_duplicates(subset=["bdns_tag"], keep="first").copy()

print(dfA.shape)
dfA.head()
```

```python
# =========================
# 3) B を読込して整形
# =========================
dfB = pd.read_csv(path_B, encoding="utf-8-sig")

# 列名の互換確保
if "bdns_tag" not in dfB.columns and "input_tag" in dfB.columns:
    dfB = dfB.rename(columns={"input_tag": "bdns_tag"})

required_cols_B = {"bdns_tag", "brick_class_candidate"}
missing = [c for c in required_cols_B if c not in dfB.columns]
if missing:
    raise ValueError(f"B に必要列が存在しない , 欠落列={missing}")

dfB["bdns_tag"] = dfB["bdns_tag"].astype(str).str.strip()
dfB["brick_class_candidate"] = dfB["brick_class_candidate"].astype(str).str.strip()

# B 側 重複削除
dfB = dfB.drop_duplicates(subset=["bdns_tag"], keep="first").copy()

# 結合キー
dfB["_join_key"] = dfB["bdns_tag"].map(normalize_tag)

print(dfB.shape)
dfB.head()
```

```python
# =========================
# 4) 結合と仕上げ
# =========================
# A を基準とした左外部結合
df = dfA.merge(
    dfB[["_join_key", "brick_class_candidate"]],
    on="_join_key",
    how="left"
)

# 最終列
df_final = df[[
    "bdns_abbriviation",
    "bdns_tag",
    "raw_ifc_class",
    "brick_class_candidate",
]].copy()

# 最終重複削除
# 完全一致行を除去
df_final = df_final.drop_duplicates(keep="first").copy()

# 並びを安定化
df_final = df_final.sort_values(
    by=["bdns_abbriviation", "bdns_tag"],
    kind="stable"
).reset_index(drop=True)

print("総レコード数 ,", len(df_final))
print("Brick 未マッチ件数 ,", df_final["brick_class_candidate"].isna().sum())
df_final.head(10)
```

```python
# =========================
# 5) 保存
# =========================
df_final.to_csv(out_csv, index=False, encoding="utf-8-sig")
df_final.to_excel(out_xlsx, index=False, engine="openpyxl")
out_csv, out_xlsx
```
