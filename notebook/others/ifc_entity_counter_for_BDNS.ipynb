エンティティ別インスタンス数の集計

```python
entity_count_dict = (
    df["ifc4_3_entity"]
    .value_counts()
    .to_dict()
)
```

```python
entity_count_dict
```

```python
def hierarchy_to_markdown_from_df(entity, entity_count_dict, level=0):
    indent = "  " * level
    entity_name = entity.name()
    count = entity_count_dict.get(entity_name, 0)

    lines = [f"{indent}- {entity_name} [instance_count={count}]"]

    for sub in entity.subtypes():
        lines.extend(
            hierarchy_to_markdown_from_df(
                sub,
                entity_count_dict,
                level + 1
            )
        )

    return lines
```

4.5 実行およびファイル出力

```python
import ifcopenshell
from ifcopenshell import ifcopenshell_wrapper

schema = ifcopenshell_wrapper.schema_by_name("IFC4X3")
ifc_root = schema.declaration_by_name("IfcRoot")

md_lines = hierarchy_to_markdown_from_df(
    ifc_root,
    entity_count_dict
)

markdown_text = "\n".join(md_lines)

with open(
    r"C:\Users\Owner\Downloads\ifc4x3_entity_instance_count_from_df.md",
    "w",
    encoding="utf-8"
) as f:
    f.write(markdown_text)
```
