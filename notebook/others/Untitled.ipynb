```python
import ifcopenshell
from ifcopenshell import ifcopenshell_wrapper

# IFC4 スキーマ定義オブジェクトを取得
schema = ifcopenshell_wrapper.schema_by_name("IFC4")

# IFC Kernel の実在する最上位 ENTITY
ifc_root = schema.declaration_by_name("IfcRoot")

def build_hierarchy(entity):
    return {
        "name": entity.name(),
        "subtypes": [
            build_hierarchy(sub)
            for sub in entity.subtypes()
        ]
    }

hierarchy = build_hierarchy(ifc_root)
```

```python
import json

with open("ifc_entity_hierarchy_ifc4.json", "w", encoding="utf-8") as f:
    json.dump(hierarchy, f, ensure_ascii=False, indent=2)
```

```python
def hierarchy_to_markdown(entity, level=0):
    indent = "  " * level
    lines = [f"{indent}- {entity['name']}"]
    for sub in entity["subtypes"]:
        lines.extend(hierarchy_to_markdown(sub, level + 1))
    return lines
```

```python
md_lines = hierarchy_to_markdown(hierarchy)
markdown_text = "\n".join(md_lines)

with open("ifc_entity_hierarchy_ifc4.md", "w", encoding="utf-8") as f:
    f.write(markdown_text)
```
