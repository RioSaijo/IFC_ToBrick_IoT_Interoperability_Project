4.1 初期化とスキーマ取得

```python
import ifcopenshell
from ifcopenshell import ifcopenshell_wrapper

# IFC ファイルをロード
ifc_path = r"C:\Users\Owner\Downloads\☆14.10階BIM整合図(2024形式).ifc"
ifc_file = ifcopenshell.open(ifc_path)

# IFC2X3 スキーマ定義を取得
#schema = ifcopenshell_wrapper.schema_by_name("IFC2X3")
# IFC4 スキーマ定義を取得
schema = ifcopenshell_wrapper.schema_by_name("IFC4")

# Kernel 実在最上位 ENTITY
ifc_root = schema.declaration_by_name("IfcRoot")
```

4.3 実インスタンス数集計

```python

def count_instances_for_entity(entity, ifc_file):
    return len(ifc_file.by_type(entity.name()))
```

4.4 Markdown 集計表現生成

```python
def hierarchy_to_markdown(entity, ifc_file, level=0):
    indent = "  " * level
    count = len(ifc_file.by_type(entity.name()))
    lines = [f"{indent}- {entity.name()} [instance_count={count}]"]

    for sub in entity.subtypes():
        lines.extend(hierarchy_to_markdown(sub, ifc_file, level + 1))

    return lines
```

4.5 実行およびファイル出力

```python
md_lines = hierarchy_to_markdown(ifc_root, ifc_file)
markdown_text = "\n".join(md_lines)

with open(r"C:\Users\Owner\Downloads\☆14.10階BIM整合図(2024形式)_instance_count_by_entity.md", "w", encoding="utf-8") as f:
    f.write(markdown_text)
```
