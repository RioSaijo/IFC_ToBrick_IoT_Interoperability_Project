```python
# 必要ライブラリのインポートとパス設定
import sys
from pathlib import Path

# プロジェクトルートを指定
PROJECT_ROOT = Path(r"C:\Users\Owner\Documents\25952501_RioSaijo\IFC_ToBrick_IoT_Interoperability_Project")

# src を sys.path に追加してトップレベルパッケージを解決
SRC_PATH = PROJECT_ROOT / "src"
if str(SRC_PATH) not in sys.path:
    sys.path.append(str(SRC_PATH))

# トポロジリゾナのインポート
from topology_reasoner.reasoner import TopologyReasoner
```

```python
from pathlib import Path

# 入力 IFC と出力 TTL のパス
IFC_PATH = PROJECT_ROOT / r"data\In\ifc\raw\51M-SKA-00-00-A-ARC-0001_IFC 2x3-P01.ifc"
OUT_TTL  = PROJECT_ROOT / r"C:\Users\Owner\Documents\25952501_RioSaijo\IFC_ToBrick_IoT_Interoperability_Project\data\intermediate\building_topology.ttl"

print("IFC:", IFC_PATH)
print("TTL:", OUT_TTL)
```

```python
# ベース名前空間は Morgate51 用の m51 を使用
BASE_NS = "https://id.morgate51.org/project#"

tr = TopologyReasoner(base_ns=BASE_NS)
model = tr.load_ifc(str(IFC_PATH))
g = tr.extract_and_map(model)

# 出力ディレクトリの作成とTTL保存
OUT_TTL.parent.mkdir(parents=True, exist_ok=True)
g.serialize(destination=str(OUT_TTL), format="turtle")
print("Wrote:", OUT_TTL)
```
